```python
# shop == 0 and no of trsn > 0 
anamoly = df[(df['noOfRcs'] == 0)& (df['noOfTrans'] > 0)].count()
print(anamoly.shape[0])
anamoly.head()
anamoly.isnull().sum()
```

```python
df['totalAmount'] = df['totalAmount'].abs()
```

```python
df['is_overflow'] = df['noOfTrans']>df['noOfRcs']
```

```python
df['total_rice'] = df['riceAfsc'] + df['riceFsc'] + df['riceAap']
```

```python
df['total_rice']
```

```python
# total_rice, wheat, sugar, rgdal,  kerosene, and salt
df['total_volume'] = df['total_rice'] + df['wheat'] +df['sugar'] +df['rgdal'] +df['kerosene'] + df['salt']
```

```python
df['trans_rate'] = np.where(df['noOfRcs'] == 0,0,(df['noOfTrans'] / df['noOfRcs']) * 100)
```

```python
df['trans_rate'].sample(5)
```

```python
def total(totalAmount):
    if totalAmount > 15000:
        return 'High Performer'
    elif totalAmount < 5000 and totalAmount > 15000:
        return 'Average Performent'
    else:
        return 'Low Performer'
df['shop_category'] = df['totalAmount'].map(total)
```

```python
district_summary = df.groupby('distName').agg({
    'totalAmount' : 'sum',
    'noOfTrans' : 'mean',
    'total_volume': 'sum'
})
```

```python
top_districts = district_summary.sort_values(
    by='totalAmount',
    ascending=False
).head(5)

print(top_districts)
```
